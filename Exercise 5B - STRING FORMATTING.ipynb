Exercise 2 - Payroll System

```python
# Payroll System

# Variables
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000.00
bonus = 35000.00
tax_rate = 7.5

# Calculations
gross_salary = basic_salary + bonus
tax_amount = gross_salary * (tax_rate/100)
net_salary = gross_salary - tax_amount

# Display Output
print("========================================")
print("PAYROLL SUMMARY")
print("========================================")
print(f"Employee:{employee_name}")
print(f"Department :{department}")
print()
print(f"Basic Salary : #{basic_salary:,.2f}")
print(f"Bonus : #{bonus:,.2f}")
print(f"Gross Salary : #{gross_salary:,.2f}")
print(f"Tax ({tax_rate}%) : #{tax_amount:,.2f}")
print(f"Net Salary : #{net_salary:,.2f}")
print("=========================================")
```

```python
# Variables
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500.00

# Calculate total cost
total_cost = quantity * unit_price

# Apply 10% discount if more than 2 items are bought
if quantity > 2:
    discount = total_cost * 0.10
else:
    discount = 0

final_amount = total_cost-discount

# Display Receipt
print("=====================================")
print("   ONLINE STORE RECEIPT")
print("=====================================")
print(f"Customer : {customer_name}")
print(f"Product  : {product}")
print(f"Quantity : {quantity}")
print(f"Unit Price : #{unit_price:,.2f}")
print(f"Total Cost : #{total_cost:,.2f}")
print(f"Discount : #{discount:,.2f}")
print(f"Final Amount : #{final_amount:,.2f}")
print("Thank you for shopping with us!")
print("======================================")
```
